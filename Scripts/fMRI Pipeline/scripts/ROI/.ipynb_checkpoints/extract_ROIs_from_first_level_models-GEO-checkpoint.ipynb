### Locate first level models 

```python
model_dir = {
    'geo' : [['IMAGE','/data00/projects/geoscan_v2/data/bids_data/derivatives_nocorrection/nipype/task-image_model-GEO-condition/']]
#    'geo' : [['IMAGE','/data00/projects/geoscan_v2/data/bids_data/derivatives_nocorrection/nipype/task-image_model-GEO_condition/'],['BANNER','/data00/projects/megameta/project1/derivatives/nipype/model_BANNER_betaseries/']]

}
```

```python
def binarize_roi(roi_path):
    '''
    Load an ROI and transform to binary format 
    '''
    
    roi_img = image.load_img(roi_path)
    roi_img_bin_data = np.where(roi_img.get_fdata()>0,1.0,0.0)
    roi_img_bin = image.new_img_like(roi_img, roi_img_bin_data)

    return roi_img_bin

# Binarize ROIs
roi_img_bin = {}
for roi in roi_dict.keys():
    roi_img_bin[roi]=binarize_roi(roi_dict[roi])


def resample_roi(roi_img_bin, resample_affine):
    '''
    resample image
    '''
        
    roi_img_rs = image.resample_to_img(roi_img_bin, resample_affine, interpolation='nearest')
    return roi_img_rs
```

```python
model_dir.items()
```

```python
subj_tmpl='{}/{}'
```

```python
spm_file
```

```python
data = []
for project, task_and_path in model_dir.items():
    
    for task_list in task_and_path:
        task=task_list[0]
        path=task_list[1]

        subjs_to_process=[s for s in os.listdir(path)]
        subjs_to_process=subjs_to_process[2:]
        print(project,': N=',len(subjs_to_process))
        print(task)

        # Resample ROIs to example con file
        model_path = subj_tmpl.format(path,subjs_to_process[3])
        spm_file = os.path.join(model_path,'SPM.mat')
        spm_mat=sio.loadmat(spm_file, 
                   squeeze_me=True, struct_as_record=False)
        contrast_dict={c.name: 'con_{:04}.nii'.format(i) for i,c in enumerate(spm_mat['SPM'].xCon,1)}
        ec=contrast_dict.items()
        ec=iter(ec)
        example_contrast=next(ec)
        con_path=os.path.join(model_path,example_contrast[1])
        con_img = image.load_img(con_path)

        rs_roi_img = {}
        for roi in roi_img_bin.keys():
            rs_roi_img[roi]=resample_roi(roi_img_bin[roi],con_img)

            #print('ROI dimension: {}'.format(rs_roi_img[roi].get_data().shape))
            #print('Con dimension: {}'.format(con_img.get_data().shape))

        for subj in subjs_to_process:
            print('Processing', subj)
            model_path = subj_tmpl.format(path,subj)

            if not os.path.exists(model_path):
                print('NO FIRST LEVEL MODEL - skipping ROI extraction')
                continue

            # load spm.mat file and create contrast dictionary    
            spm_file = os.path.join(model_path,'SPM.mat')
            spm_mat=sio.loadmat(spm_file, 
                   squeeze_me=True, struct_as_record=False)
            contrast_dict={c.name: 'con_{:04}.nii'.format(i) for i,c in enumerate(spm_mat['SPM'].xCon,1)}

            # loop through contrasts 
            for (name,con) in contrast_dict.items():
                con_path=os.path.join(model_path,con)
                con_img=image.load_img(con_path)
                row = {'participant_id': subj,'study':project,'task':task,'contrast':name}

                #loop over rois
                for roi in rs_roi_img:
                    #print('\textracting data from {} roi'.format(roi))
                    intersect_data=con_img.get_fdata() * rs_roi_img[roi].get_fdata()
                    roidata=np.nanmean(intersect_data)
                    row[roi]=roidata
                data.append(row)
        
    roi_df=pd.DataFrame(data)
    roi_df.to_csv('roi_metaanalyses_geo_060525.csv', index=False)
```
