```python
import os
import json
import pandas as pd
from pathlib import Path
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# Import template system modules
from json_template_system import (
    create_example_templates, 
    scan_project_for_subjects,
    AnalysisTemplateManager,
    load_project_config
)

print("✅ All imports successful!")
print(f"📅 Setup started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
# 🔧 EDIT THESE PATHS FOR YOUR PROJECT
PROJECT_PATH = "/data00/projects/geoscan_v2"  # Your main project directory
TEMPLATES_DIR = "./analysis_templates"        # Where to store template files
PROJECT_NAME = "geoscan_v2"                   # Name for your project
TASK_NAME = "geoscan"                         # fMRI task name
SESSIONS = ["ses-t2", "ses-t3"]               # Session names to combine

# Create templates directory
os.makedirs(TEMPLATES_DIR, exist_ok=True)

print(f"📁 Project path: {PROJECT_PATH}")
print(f"📋 Templates directory: {TEMPLATES_DIR}")
print(f"🎯 Task: {TASK_NAME}")
print(f"📊 Project: {PROJECT_NAME}")
print(f"📅 Sessions: {', '.join(SESSIONS)}")

# Verify project path exists
if os.path.exists(PROJECT_PATH):
    print(f"✅ Project directory found")
else:
    print(f"⚠️ Project directory not found - please check path")
```

```python
print("📋 Creating example templates...")

example_templates = create_example_templates(TEMPLATES_DIR)

print(f"\n✅ Example templates created:")
for template_type, template_path in example_templates.items():
    file_size = os.path.getsize(template_path) if os.path.exists(template_path) else 0
    print(f"  📄 {template_type}: {os.path.basename(template_path)} ({file_size} bytes)")

print(f"\n💡 Next steps:")
print(f"  1. Review and edit the templates below")
print(f"  2. Update paths to match your project structure")
print(f"  3. Customize analysis parameters as needed")
```

```python
manager = AnalysisTemplateManager(TEMPLATES_DIR)
templates = manager.list_templates()

print(f"📋 Available Templates ({len(templates)} total):")
print("=" * 50)

for template in templates:
    print(f"📄 {template['name']}")
    print(f"   Type: {template['type']}")
    print(f"   Created: {template['created']}")
    print(f"   Description: {template.get('description', 'No description')}")
    print()
```

```python
project_templates = [t for t in templates if t['type'] == 'project']

if project_templates:
    project_template = manager.load_template(project_templates[0]['path'])
    
    print("📊 PROJECT TEMPLATE SUMMARY:")
    print("=" * 35)
    print(f"Project Name: {project_template.get('project_name')}")
    print(f"Project Path: {project_template.get('project_path')}")
    print(f"Task Name: {project_template.get('task_name')}")
    print(f"Sessions: {', '.join(project_template.get('sessions', []))}")
    print(f"BIDS Compliant: {project_template.get('bids_compliant')}")
    print(f"Description: {project_template.get('description', 'None')}")
    
    # Show full template for editing
    print(f"\n📋 Full Project Template (edit as needed):")
    print(json.dumps(project_template, indent=2))
    
    # Check if paths match your configuration
    template_path = project_template.get('project_path')
    template_task = project_template.get('task_name')
    
    print(f"\n🔍 Template Validation:")
    if template_path == PROJECT_PATH:
        print(f"  ✅ Project path matches configuration")
    else:
        print(f"  ⚠️ Project path mismatch:")
        print(f"     Template: {template_path}")
        print(f"     Config: {PROJECT_PATH}")
    
    if template_task == TASK_NAME:
        print(f"  ✅ Task name matches configuration")
    else:
        print(f"  ⚠️ Task name mismatch:")
        print(f"     Template: {template_task}")
        print(f"     Config: {TASK_NAME}")
        
else:
    print("❌ No project templates found")
```

```python
UPDATE_PROJECT_TEMPLATE = True  # Set to True to update

if UPDATE_PROJECT_TEMPLATE and project_templates:
    print("🔧 Updating project template with your configuration...")
    
    # Load current template
    project_template = manager.load_template(project_templates[0]['path'])
    
    # Update with your settings
    project_template.update({
        'project_name': PROJECT_NAME,
        'project_path': PROJECT_PATH,
        'derivatives_path': f"{PROJECT_PATH}/data/bids_data/derivatives_nocorrection",
        'task_name': TASK_NAME,
        'sessions': SESSIONS,
        'modified': datetime.now().isoformat(),
        'description': f"Updated template for {PROJECT_NAME} analysis"
    })
    
    # Save updated template
    updated_file = manager.create_template('project', project_template, 
                                         f'project_{PROJECT_NAME}_updated.json')
    
    print(f"✅ Updated project template saved: {updated_file}")
    print(f"\n📋 Updated settings:")
    print(f"  Project: {project_template['project_name']}")
    print(f"  Path: {project_template['project_path']}")
    print(f"  Task: {project_template['task_name']}")
    print(f"  Sessions: {', '.join(project_template['sessions'])}")
    
else:
    print("⏭️ Project template update skipped")
```

```python
SCAN_FOR_SUBJECTS = True  # Set to True to scan

if SCAN_FOR_SUBJECTS:
    print("🔍 Auto-scanning project for subjects...")
    
    # Use updated project template if available
    project_templates = manager.list_templates('project')
    if project_templates:
        # Use most recent template
        latest_template = sorted(project_templates, key=lambda x: x['created'])[-1]
        project_template = manager.load_template(latest_template['path'])
        print(project_template)
        print(f"Using project template: {latest_template['name']}")
        
        # Auto-scan for subjects
        dataset_template = scan_project_for_subjects(project_template, TASK_NAME)
        
        if dataset_template and dataset_template.get('subjects'):
            # Save auto-generated dataset
            dataset_file = manager.create_template('dataset', dataset_template, 
                                                 f'dataset_{PROJECT_NAME}_auto.json')
            
            n_subjects = len(dataset_template['subjects'])
            print(f"\n✅ Auto-scan complete!")
            print(f"📊 Found {n_subjects} subjects with required sessions")
            print(f"📄 Dataset template saved: {dataset_file}")
            
            # Show summary of found subjects
            print(f"\n👥 Subject Summary:")
            for i, subject in enumerate(dataset_template['subjects'][:10]):
                subject_id = subject['subject_id']
                sessions = list(subject.get('sessions', {}).keys())
                print(f"  {i+1:2d}. {subject_id}: {', '.join(sessions)}")
            
            if n_subjects > 10:
                print(f"  ... and {n_subjects - 10} more subjects")
                
        else:
            print(f"❌ No subjects found during auto-scan")
            print(f"💡 Check your project structure and file naming")
            
    else:
        print(f"❌ No project template found for scanning")
        
else:
    print("⏭️ Subject scanning skipped")
```

```python
analysis_templates = [t for t in manager.list_templates() if t['type'] == 'analysis']

if analysis_templates:
    analysis_template = manager.load_template(analysis_templates[0]['path'])
    
    print("🔧 ANALYSIS TEMPLATE SUMMARY:")
    print("=" * 35)
    print(f"Analysis Name: {analysis_template.get('analysis_name')}")
    print(f"Analysis Type: {analysis_template.get('analysis_type')}")
    print(f"Description: {analysis_template.get('description')}")
    
    # Parameters
    params = analysis_template.get('parameters', {})
    print(f"\n⚙️ Key Parameters:")
    print(f"  TR: {params.get('tr', 'Not set')} seconds")
    print(f"  High-pass filter: {params.get('hpf', 'Not set')} seconds")
    print(f"  Smoothing FWHM: {params.get('smoothing', 'Not set')} mm")
    print(f"  Mask threshold: {params.get('mask_threshold', 'Not set')}")
    print(f"  Model basis: {params.get('model_basis', 'Not set')}")
    
    # Conditions
    conditions = analysis_template.get('conditions', {})
    main_conditions = conditions.get('main_conditions', [])
    modulators = conditions.get('parametric_modulators', [])
    
    print(f"\n🎯 Task Conditions:")
    if main_conditions:
        print(f"  Main conditions: {', '.join(main_conditions)}")
    else:
        print(f"  Main conditions: Will auto-detect from events files")
    
    if modulators:
        print(f"  Parametric modulators: {len(modulators)}")
        for mod in modulators:
            condition = mod.get('condition', 'unknown')
            modulator = mod.get('modulator_name', 'unknown')
            print(f"    - {modulator} for {condition}")
    
    # Contrasts
    contrasts = analysis_template.get('contrasts', [])
    print(f"\n🔄 Contrasts ({len(contrasts)} defined):")
    for i, contrast in enumerate(contrasts, 1):
        name = contrast.get('name', f'contrast_{i}')
        contrast_type = contrast.get('type', 'unknown')
        print(f"  {i}. {name} ({contrast_type.upper()}-contrast)")
        
        if contrast_type == 't' and 'vector' in contrast:
            print(f"     Vector: {contrast['vector']}")
        elif contrast_type == 'f' and 'matrix' in contrast:
            matrix = contrast['matrix']
            print(f"     Matrix: {len(matrix)}x{len(matrix[0]) if matrix else 0}")
    
else:
    print("❌ No analysis templates found")
```
