# README

- This script contains a series of functions to clean up our geodata a little bit. At the moment, it only works if the participant's unzipped geodata is in the Timeline/3_S3 folder, or a completed participant has data in the Timeline/2_S2 folder. 

- This script has three main functions, one to trim the location data to the study period, and one that converts this into .csvs and splits it by study period. There is one more function to combine all of the new csvs. 

### To update the geodata, do the following:
- Update redcap data (use get_redcap_data(download=TRUE) from GeoRemote/Scripts/Utility Scripts/geo_redcap.R)
- Run the following cell to import geo_codebase.py
- Run trim_and_save() to save trimmed geodata
- Run trim_and_save_by_study_period(study_period) for 'baseline' and 'intervention' to save .csvs
- Run merge_geodata(study_period) for 'baseline' and 'intervention' to save merged data into one file

### trim_and_save()
-   Data is pulled from GeoRemote/Timeline/3_S3, and then GeoRemote/Timeline/2_S2.
-   pID is taken from the folder using extract_participant_id(timeline_folder)
-   If the file already exists in the output folder, that ppt is skipped.
-   The json is read and only objects that fall within the redcap dates are used.
-   Start date is found using these dates (whichever exists) in order: ["screenb_date", "s1_delivery","consent_date"]
-   After 3_S3 is completed, the function loops through 2_S2, and skips participants if they are completed or if they already exist in the output function.

### trim_and_save_by_study_period(study_period)
-   arguments: 'baseline' or 'intervention'
-   Pulls data from GeoRemote/Data/Geodata/clean/trimmed_geodata/
-   filter_and_write_geodata_by_study_period(geodata_file, dates, output_file,chunk_size) is where the majority of processing happens. Processing includes:
    -   trimming geodata by study period
    -   data saved from json include ["latitudeE7", "longitudeE7", "altitude", "velocity", "timestamp", "heading"]
    -   renamed columns: {'latitudeE7': 'lat', 'longitudeE7': 'lon', "timestamp": "datetime"}
    -   lat and lon get '.' added to it in the second index
    -   filename and pid columns added (participant id)
